```python
# Select numerical columns
Num = Ss.select_dtypes(include=np.number)

# Set general style
sns.set(style="whitegrid")

# Loop through numerical features
for col in Num:
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # Boxplot: Numerical vs Target
    sns.boxplot(
        x='Quality of Sleep',
        y=col,
        data=Ss,
        ax=axes[0]
    )
    axes[0].set_title(f'{col} vs Quality of Sleep (Boxplot)')

    # KDE: Distribution by Target
    sns.kdeplot(
        data=Ss,
        x=col,
        hue='Quality of Sleep',
        fill=True,
        common_norm=False,
        alpha=0.5,
        ax=axes[1]
    )
    axes[1].set_title(f'{col} Distribution by Quality of Sleep ')

    plt.tight_layout()
    plt.show()
```
